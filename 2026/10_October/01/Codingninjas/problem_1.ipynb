## Problem statement

A factory floor uses a mechanical tally counter to track the total number of units produced on a production line. The counter's display is a row of digit wheels, read left to right, so its current reading is given as an array count of single digits.



You are given the array count and an integer produced, the number of additional units manufactured during the latest shift. Return the new counter reading as an array of digits, in the same left-to-right order, after adding produced to the value shown by count.

Detailed explanation ( Input/output format, Notes, Images )

### Sample Input 1:
```
4 7 9 3
258
```


### Sample Output 1:
```
5 0 5 1
```


### Explanation:

The counter currently reads 4793. Producing 258 more units during the shift gives 4793 + 258 = 5051, so the display becomes the digits 5, 0, 5, 1.



### Sample Input 2:
```
9 9 9
1
```



### Sample Output 2:
```
1 0 0 0
```


### Explanation:

The counter reads 999. Producing 1 more unit rolls every wheel over at once: 999 + 1 = 1000. A new leading wheel appears, so the display becomes the digits 1, 0, 0, 0.


```
Expected Time Complexity: O(n)

Expected Auxiliary Space: O(n)

```
### Constraints:

- 1 ≤ count.length ≤ 10⁴

- 0 ≤ count[i] ≤ 9

- count does not contain any leading zeros, except when count itself represents the value zero.

- 1 ≤ produced ≤ 10⁴

- but this code has a problem, 

Your solution fails because Python 3.11+ enforces an integer string conversion limit (`sys.set_int_max_str_digits`, default: **4,300 digits**).

The problem constraints specify `count.length <= 10^4` (up to 10,000 digits). When converting a string of more than 4,300 digits with `int(x)`, Python raises:

```text
ValueError: Exceeds the limit (4300) for integer string conversion

```

In addition, building strings repeatedly via `x += str(i)` causes $O(n^2)$ time overhead.

---

### Optimal $O(n)$ Solution (Digit-by-Digit Addition)

Treat `produced` as a running carry and add it directly from right to left (like manual column addition):

```python
def tallyWheelUpdate(count, produced):
    result = []
    i = len(count) - 1
    carry = produced

    # Traverse from rightmost digit, adding carry as you go
    while i >= 0 or carry > 0:
        if i >= 0:
            carry += count[i]
            i -= 1

        result.append(carry % 10)
        carry //= 10

    # Digits were appended in reverse order (ones, tens, etc.)
    result.reverse()
    return result

```

### Complexity

* **Time Complexity:** $O(n)$ — single pass over the array and remaining carry digits.
* **Auxiliary Space:** $O(n)$ — creates the output array without large string conversions.
