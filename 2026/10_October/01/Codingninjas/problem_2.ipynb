### Problem statement
You are given an 8 x 8 grid representing the floor of a harbor control room. There is exactly one searchlight tower marked 'L'. Some cells contain stacked crates marked 'C', and some cells contain smugglers' boats marked 'b'. Every other cell is empty and marked '.'.



The searchlight can sweep in a straight beam in any of the four directions (up, down, left, right) from the tower, continuing until it exits the grid or reaches another object. The beam spots a boat if that boat is the first object the beam reaches while sweeping in some direction.



Note that the beam cannot pass through a crate. If a crate lies between the tower and a boat in the same direction, that boat is not spotted, even if it would otherwise be visible.

Return the number of boats spotted by the searchlight.

. . . . . . . .

3



Explanation:
